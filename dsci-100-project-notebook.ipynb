#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
march_data %>%
  count(tourney_finish)
# Select relevant variables for the research question
march_tidy <- march_data %>%
  select(year, school, seed, reg_wins_pct, bid, tourney_finish) %>%

  # Create a categorical outcome indicating Sweet Sixteen qualification
  mutate(
    reached_sweet16 = if_else(
      tourney_finish %in% c(
        "top_16_loss",
        "top_8_loss",
        "top_4_loss",
        "top_2_loss",
        "champ"
      ),
      "Yes",
      "No"
    ),
    reached_sweet16 = as_factor(reached_sweet16)
  ) %>%

  # Remove the original tournament outcome column
  select(-tourney_finish)

# Display the first six rows of the tidy dataset
head(march_tidy, 6)
```

```r
# Plot 1: Distribution of regular-season winning percentage

ggplot(march_tidy, aes(x = reg_wins_pct)) +
  geom_histogram(
    binwidth = 5,
    boundary = 0,
    fill = "steelblue",
    color = "white"
  ) +
  scale_x_continuous(
    breaks = seq(40, 100, by = 10)
  ) +
  labs(
    title = "Distribution of Regular-Season Winning Percentage",
    subtitle = "NCAA Women's Basketball Tournament (1982–2018)",
    x = "Regular-Season Winning Percentage (%)",
    y = "Number of Teams"
  ) +
  theme_minimal(base_size = 13)
```

```r
# Calculate the proportion of teams in each category
sweet16_proportions <- march_tidy %>%
  count(reached_sweet16) %>%
  mutate(
    proportion = n / sum(n),
    reached_sweet16 = factor(
      reached_sweet16,
      levels = c("No", "Yes")
    )
  )

# Plot 2: Relative proportions of Sweet Sixteen qualification
ggplot(
  sweet16_proportions,
  aes(x = reached_sweet16, y = proportion)
) +
  geom_col(
    fill = "steelblue",
    width = 0.6
  ) +
  geom_text(
    aes(label = scales::percent(proportion, accuracy = 0.1)),
    vjust = -0.5,
    size = 4.5
  ) +
  scale_y_continuous(
    labels = scales::percent_format(accuracy = 1),
    expand = expansion(mult = c(0, 0.12))
  ) +
  labs(
    title = "Proportion of Teams Reaching the Sweet Sixteen",
    subtitle = "NCAA Women's Basketball Tournament (1982–2018)",
    x = "Reached Sweet Sixteen",
    y = "Percentage of Teams"
  ) +
  theme_minimal(base_size = 13)
```

One limitation of KNN is that its predictions are sensitive to the number of neighbors (K) and the scales of the predictors. Additionally, the dataset is imbalanced, as only approximately 28.3% of teams reached the Sweet Sixteen. Therefore, overall accuracy alone may not adequately reflect model performance.

I plan to split the data into 75% training and 25% testing sets using stratified sampling to preserve the class proportions. Numeric predictors will be standardized, while the categorical predictor, bid type, will be converted into dummy variables. All preprocessing will be estimated using the training data to avoid data leakage.

I will use five-fold cross-validation on the training set to select an appropriate K value. The final model will then be evaluated on the untouched testing set using accuracy, recall, balanced accuracy, and a confusion matrix. Recall is particularly important for measuring how well the model identifies teams that actually reach the Sweet Sixteen, while balanced accuracy accounts for differences in class frequencies.
